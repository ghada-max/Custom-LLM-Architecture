```python
import torch
import torch.nn as nn
class TinyModel(nn.Module):
  def __init__(self):
    super().__init__()
#x-->weight*x + b -->prediction-->compare with target-->loss
    self.w=nn.Parameter(torch.tensor([2.0]))
    self.b=nn.Parameter(torch.tensor([0.0]))

  def forward(self,x):
     return (self.w * x ) + self.b

```

```python
my_model = TinyModel()
print("w:", my_model.w)
print("b:", my_model.b)
my_input=torch.tensor([3.0])
prediction=my_model(my_input)
print(prediction)
target=18
```

```python

target = 18
model = TinyModel()
learning_rate = 0.0002
x = torch.tensor([3.0])

for step in range(1000):
    prediction = model(x)
    loss = (prediction - target) ** 2

    model.zero_grad()
    loss.backward()

    print(
        f"step={step}",
        f"prediction={prediction.item():.2f}",
        f"loss={loss.item():.2f}",
        f"w.grad={model.w.grad.item():.2f}",
    )
    #Learning-rate or overshting
    #update parameters
    with torch.no_grad():
        model.w -=learning_rate*model.w.grad
        model.b -=learning_rate*model.b.grad

    #Clear old gradient
    model.w.grad.zero_()
    model.b.grad.zero_()
```

```python
#Embedding
import torch
import torch.nn as nn
vocab_size=24_000
dim=768
class CustomModel(nn.Module):
  def __init__(self,vocab_size:int,dim:int):
     super().__init__()
     #tokenid-->vector of 768 numbers
     self.embedding=nn.Embedding(vocab_size,dim)
     #normalization
     self.norm=nn.LayerNorm(dim)
     #FeedForward
     self.ff_head=nn.Linear(dim,vocab_size)
  def forward(self,token_ids):
    embed_output=self.embedding(token_ids)
    embed_output=self.norm(embed_output)
    logits=self.ff_head(embed_output)
    return logits

    #create
    my_model=CustomModel(
        vocab_size=24_000,
        dim=768
    )
token_ids=torch.tensor([
    [10,25,98,16],
    [4,77,31,9]
])
token_logits=my_model(token_ids)
```
